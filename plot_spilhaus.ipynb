```julia
xx = rand(100) .* 360. .- 180.
yy = rand(100) .* 180. .- 90.
zz = xx .* xx .+ xx .* yy
```

```julia
using CairoMakie, GeoMakie

fig = Figure(size = (1000, 1000))
ga = GeoAxis(
    fig[1, 1]; # any cell of the figure's layout
    dest = "+proj=spilhaus", 
    xgridvisible = false, 
    ygridvisible = false,
    limits = ((-180, 180.), (-85., 85.))
    # the CRS in which you want to plot
)
GeoMakie.land(ga)
# You can plot your data the same way you would in Makie
scatter!(ga, xx, yy; color = zz, strokecolor = (:black, 0.2))
fig
```
